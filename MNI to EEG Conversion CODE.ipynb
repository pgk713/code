```python
import numpy as np
import mne
from mne.channels import make_standard_montage
from mne.transforms import apply_trans, invert_transform

# Your MNI coordinates (in mm)
mni_coords = np.array([45, 51, -12])  # Replace with your coordinates

# Load standard montage (10/10 system)
montage = make_standard_montage('standard_1020')

# Get the EEG channel positions in the head coordinate space
ch_coords = montage.get_positions()['ch_pos']
ch_names = list(ch_coords.keys())

# Convert the coordinates to a numpy array
ch_coords_array = np.array([ch_coords[ch_name] for ch_name in ch_names])

# Load the MNI template to get the transform from MNI to head coordinates
fsaverage = mne.datasets.fetch_fsaverage()
trans = mne.coreg.estimate_head_mri_t('fsaverage', fsaverage)

# Invert the transform to go from MNI to head space
mni_to_head_trans = invert_transform(trans)

# Apply the inverse transform to your MNI coordinates
head_coords = apply_trans(mni_to_head_trans, mni_coords)

# Compute the Euclidean distances between the head coordinates and each channel position
distances = np.linalg.norm(ch_coords_array - head_coords, axis=1)

# Find the indices of the 3 closest channels
top_3_idx = np.argsort(distances)[:3]

# Get the nearest 3 EEG labels
top_3_labels = [ch_names[idx] for idx in top_3_idx]

print(f"The top 3 nearest EEG labels for MNI coordinates {mni_coords} are {top_3_labels}.")
```
