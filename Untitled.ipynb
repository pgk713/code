```python

import numpy as np
import pandas as pd

from ISLP import load_data, confusion_table


from sklearn.discriminant_analysis import (LinearDiscriminantAnalysis as LDA,
                                           QuadraticDiscriminantAnalysis as QDA)
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier

pd.set_option("display.max_columns", None)
```

```python
# load data
Smarket = load_data('Smarket')
Smarket.head()

Smarket.shape
Smarket.columns

Smarket.describe()

```

```python

X = Smarket[['Lag1', 'Lag2']]
L = Smarket.Direction  
```

```python
# ------------- LDA
lda = LDA(store_covariance=True) #keep the fitted covariance matrix
lda.fit(X, L)

lda.classes_  # double check the order of the class

# prior P(y=k)
lda.priors_

# the mean predictor value in each class
pd.DataFrame(lda.means_, index=lda.classes_, columns=X.columns)

# the common covariance across all classes
lda.covariance_

# the LDA linear decision boundary
lda.coef_
lda.intercept_
# -0.07*Lag1 - 0.04*Lag2 + 0.07 > 0, predict "Up" (the second class)

# prediction results
lda_pred = lda.predict(X)
print(lda_pred)
print(confusion_table(lda_pred, L))

# Overall test accuracy
np.mean(lda_pred == L)

# if want to double check
# Smarket["Lag1"].iloc[0]*lda.coef_[0,0] + Smarket["Lag2"].iloc[0]*lda.coef_[0,1] + lda.intercept_
# lda.decision_function(X)[0]
# lda_pred[0]



```

```python

# ----------- QDA
qda = QDA(store_covariance=True) #keep the fitted covariance matrix
qda.fit(X, L)

qda.classes_
qda.priors_

# the mean predictor value in each class (the same as LDA)
pd.DataFrame(qda.means_, index=qda.classes_, columns=X.columns)

# covariance for "Down" group
print(qda.covariance_[0])

# "Up group"
print(qda.covariance_[1])

# predictions
qda_pred = qda.predict(X)
print(qda_pred)
print(confusion_table(qda_pred, L))

# accuracy
print(np.mean(qda_pred == L))

# no straightforward decision boundary like LDA (but the same rule applied)
qda.decision_function(X)[0:3]
qda_pred[0:3]


```

```python
# ---------- Naive Bayes

NB = GaussianNB()
NB.fit(X, L)

NB.classes_
NB.class_prior_

# means and variances per group (no covariance. Why? assumes no covariance)
pd.DataFrame(NB.theta_, index=NB.classes_, columns=X.columns)
pd.DataFrame(NB.var_, index=NB.classes_, columns=X.columns) 

nb_pred = NB.predict(X)
print(nb_pred)
print(confusion_table(nb_pred, L))

np.mean(nb_pred == L)


```

```python
# ---------- KNN

X_arr = np.asarray(X)

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_arr, L)
knn_pred = knn.predict(X_arr)
confusion_table(knn_pred, L)

np.mean(knn_pred == L)
```

```python
# But it is biased if we use the training data to assess its performance!
# the observation itself is considered one of its nearest neighbor

X_train, X_test = X.loc[Smarket.Year < 2005], X.loc[Smarket.Year == 2005]
L_train, L_test = L.loc[Smarket.Year < 2005], L.loc[Smarket.Year == 2005]

X_train_arr, X_test_arr = np.asarray(X_train), np.asarray(X_test)

knn1 = KNeighborsClassifier(n_neighbors=3)
knn1.fit(X_train_arr, L_train)


knn1_pred = knn1.predict(X_test_arr)
confusion_table(knn1_pred, L_test)
np.mean(knn1_pred == L_test)
```
